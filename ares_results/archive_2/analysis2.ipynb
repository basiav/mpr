# Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Load Data

```python
udf_raw = pd.read_csv("results_uni.csv", sep=';', header=None, names=["algo", "threads", "t", "t1", "t2", "t3", "t4", "t5"])
udf_raw.head()
```

```python
bdf_raw = pd.read_csv("results.csv", sep=';', header=None, names=["algo", "buckets", "t", "t1", "t2", "t3", "t4", "t5"])
bdf_raw.head()
```

```python
odf_raw = pd.read_csv("results_oriented.csv", sep=';', header=None, names=["algo", "threads", "buckets", "t", "t1", "t2", "t3", "t4", "t5"])
odf_raw.head()
```

```python
udf = udf_raw.groupby(['algo', 'threads']).agg(
    t_mean=('t', 'mean'), t_std=('t', 'std'),
    t1_mean=('t1', 'mean'), t1_std=('t1', 'std'),
    t2_mean=('t2', 'mean'), t2_std=('t2', 'std'),
    t3_mean=('t3', 'mean'), t3_std=('t3', 'std'),
    t4_mean=('t4', 'mean'), t4_std=('t4', 'std'),
    t5_mean=('t5', 'mean'), t5_std=('t5', 'std'),
    num_runs=('t', 'size')
).reset_index()

udf.head()
```

```python
bdf = bdf_raw.groupby(['algo', 'buckets']).agg(
    t_mean=('t', 'mean'), t_std=('t', 'std'),
    t1_mean=('t1', 'mean'), t1_std=('t1', 'std'),
    t2_mean=('t2', 'mean'), t2_std=('t2', 'std'),
    t3_mean=('t3', 'mean'), t3_std=('t3', 'std'),
    t4_mean=('t4', 'mean'), t4_std=('t4', 'std'),
    t5_mean=('t5', 'mean'), t5_std=('t5', 'std'),
    num_runs=('t', 'size')
).reset_index()

bdf.head()
```

```python
odf = odf_raw.groupby(['algo', 'threads', 'buckets']).agg(
    t_mean=('t', 'mean'), t_std=('t', 'std'),
    t1_mean=('t1', 'mean'), t1_std=('t1', 'std'),
    t2_mean=('t2', 'mean'), t2_std=('t2', 'std'),
    t3_mean=('t3', 'mean'), t3_std=('t3', 'std'),
    t4_mean=('t4', 'mean'), t4_std=('t4', 'std'),
    t5_mean=('t5', 'mean'), t5_std=('t5', 'std'),
    num_runs=('t', 'size')
).reset_index()

odf.head()
```

# Execution time

```python
y_logged = True

for series in udf.groupby(["algo"]):
    series_df = series[1]
    series_nm = series[0][0]
    
    plt.plot(series_df["threads"], series_df["t_mean"], label="total execution")
    plt.plot(series_df["threads"], series_df["t1_mean"], label="array fill")
    plt.plot(series_df["threads"], series_df["t2_mean"], label="distribution to buckets")
    plt.plot(series_df["threads"], series_df["t3_mean"], label="sorting buckets")
    plt.plot(series_df["threads"], series_df["t4_mean"], label="merging results")
    
    plt.title(f"{series_nm} - Execution Time")
    plt.legend()

    if y_logged:
        plt.yscale("log")
        plt.ylabel(r"Time [s] $\log_{2}$ scale")
    else:
        plt.ylabel("Time [s]")
    plt.xlabel("Number of Threads")

    plt.grid()
    plt.show()
```

```python
y_logged = True
num_runs_info = True

total_ex_times = []
array_fill_times = []
distribution_times = []
sorting_times = []
merging_times = []
merging_buckets_times_sort_3 = []
num_runs_arr = []

for series in udf.groupby(["algo"]):
    series_df = series[1]
    series_nm = series[0][0]

    total_ex_times.append(series_df["t_mean"].values)
    array_fill_times.append(series_df["t1_mean"].values)
    distribution_times.append(series_df["t2_mean"].values)
    sorting_times.append(series_df["t3_mean"].values)
    merging_times.append(series_df["t4_mean"].values)
    # Non NaN for sort_3
    merging_buckets_times_sort_3.append(series_df["t5_mean"].values)

    # Additional info
    num_runs = series_df["num_runs"].values[0]
    num_runs_arr.append(num_runs)

fig, ax = plt.subplots(3,2, figsize=(12,12))

# Total Execution Time
i, j = 0, 0
for sort_no, data in enumerate(total_ex_times):
    ax[i][j].plot(series_df["threads"], data, label=f"sort_{sort_no+1}")
ax[i][j].set_title("Total Execution Times")
ax[i][j].set_xlabel("Number of Threads")
if y_logged:
    ax[i][j].set_yscale("log")
    ax[i][j].set_ylabel(r"Time [s] $\log_{2}$ scale")
else:
    ax[i][j].set_ylabel("Time [s]")
ax[i][j].legend()

if num_runs_info:
    ax[i][j].set_title(f"Total Execution Times - mean from {num_runs_arr[0]} runs")

# Array Fill Time
i, j = 0, 1
for sort_no, data in enumerate(array_fill_times):
    ax[i][j].plot(series_df["threads"], data, label=f"sort_{sort_no+1}")
ax[i][j].set_title("Array Fill Times")
ax[i][j].set_xlabel("Number of Threads")
if y_logged:
    ax[i][j].set_yscale("log")
    ax[i][j].set_ylabel(r"Time [s] $\log_{2}$ scale")
else:
    ax[i][j].set_ylabel("Time [s]")
ax[i][j].legend()

if num_runs_info:
    ax[i][j].set_title(f"Array Fill Times - means from {num_runs_arr[1]} runs")

# Distribution to Buckets Time
i, j = 1, 0
for sort_no, data in enumerate(distribution_times):
    ax[i][j].plot(series_df["threads"], data, label=f"sort_{sort_no+1}")
ax[i][j].set_title("Distribution to Buckets Times")
ax[i][j].set_xlabel("Number of Threads")
if y_logged:
    ax[i][j].set_yscale("log")
    ax[i][j].set_ylabel(r"Time [s] $\log_{2}$ scale")
else:
    ax[i][j].set_ylabel("Time [s]")
ax[i][j].legend()

if num_runs_info:
    ax[i][j].set_title(f"Distribution to Buckets Times - means from {num_runs_arr[2]} runs")


# Sorting buckets Time
i, j = 1, 1
for sort_no, data in enumerate(sorting_times):
    ax[i][j].plot(series_df["threads"], data, label=f"sort_{sort_no+1}")
ax[i][j].set_title("Sorting Buckets Times")
ax[i][j].set_xlabel("Number of Threads")
if y_logged:
    ax[i][j].set_yscale("log")
    ax[i][j].set_ylabel(r"Time [s] $\log_{2}$ scale")
else:
    ax[i][j].set_ylabel("Time [s]")
ax[i][j].legend()

if num_runs_info:
    ax[i][j].set_title(f"Sorting Buckets Times - means from {num_runs_arr[3]} runs")


# Merging array Time
i, j = 2, 0
for sort_no, data in enumerate(merging_times):
    ax[i][j].plot(series_df["threads"], data, label=f"sort_{sort_no+1}")
    if not np.isnan(merging_buckets_times_sort_3[sort_no]).all():
        ax[i][j].plot(series_df["threads"], merging_buckets_times_sort_3[sort_no], label=f"sort_{sort_no+1} - merging buckets", linestyle='dashed')
ax[i][j].set_title("Merging Array Times")
ax[i][j].set_xlabel("Number of Threads")
if y_logged:
    ax[i][j].set_yscale("log")
    ax[i][j].set_ylabel(r"Time [s] $\log_{2}$ scale")
else:
    ax[i][j].set_ylabel("Time [s]")
ax[i][j].legend()

if num_runs_info:
    ax[i][j].set_title(f"Merging Array Times - means from {num_runs_arr[3]} runs")

# Merging Buckets Time (sort_3)
i, j = 2, 1
for sort_no, data in enumerate(merging_buckets_times_sort_3):
    ax[i][j].plot(series_df["threads"], data, label=f"sort_{sort_no+1}")
ax[i][j].set_title("Merging Buckets Times")
ax[i][j].set_xlabel("Number of Threads")
if y_logged:
    ax[i][j].set_yscale("log")
    ax[i][j].set_ylabel(r"Time [s] $\log_{2}$ scale")
else:
    ax[i][j].set_ylabel("Time [s]")
ax[i][j].legend()

if num_runs_info:
    ax[i][j].set_title(f"Merging Buckets Times - means from {num_runs_arr[3]} runs")

plt.tight_layout()
```

# Speedup

```python
for series in udf.groupby(["algo"]):
    series_df = series[1]
    series_nm = series[0][0]
    
    T1 = series_df[series_df["threads"] == 1]["t_mean"].values[0]
    T11 = series_df[series_df["threads"] == 1]["t1_mean"].values[0]
    T12 = series_df[series_df["threads"] == 1]["t2_mean"].values[0]
    T13 = series_df[series_df["threads"] == 1]["t3_mean"].values[0]
    T14 = series_df[series_df["threads"] == 1]["t4_mean"].values[0]
    
    plt.plot(series_df["threads"], T1/series_df["t_mean"], label="total execution")
    plt.plot(series_df["threads"], T11/series_df["t1_mean"], label="array fill")
    plt.plot(series_df["threads"], T12/series_df["t2_mean"], label="distribution to buckets")
    plt.plot(series_df["threads"], T13/series_df["t3_mean"], label="sorting buckets")
    plt.plot(series_df["threads"], T14/series_df["t4_mean"], label="merging results")
    
    perfect_speedup = series_df["threads"].values
    plt.plot(perfect_speedup, perfect_speedup, label="perfect speedup", linestyle='--', color='black')
    
    plt.title(f"{series_nm} - Speedup")
    plt.legend()
    plt.xlabel("Number of Threads")
    plt.grid()
    plt.ylabel("Speedup")
    plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12,12))

i, j = 0, 0

for series in udf.groupby(["algo"]):
    series_df = series[1]
    series_nm = series[0][0]
    
    T1 = series_df[series_df["threads"] == 1]["t_mean"].values[0]
    T11 = series_df[series_df["threads"] == 1]["t1_mean"].values[0]
    T12 = series_df[series_df["threads"] == 1]["t2_mean"].values[0]
    T13 = series_df[series_df["threads"] == 1]["t3_mean"].values[0]
    T14 = series_df[series_df["threads"] == 1]["t4_mean"].values[0]
    
    ax[i][j].plot(series_df["threads"], T1/series_df["t_mean"], label="total execution")
    ax[i][j].plot(series_df["threads"], T11/series_df["t1_mean"], label="array fill")
    ax[i][j].plot(series_df["threads"], T12/series_df["t2_mean"], label="distribution to buckets")
    ax[i][j].plot(series_df["threads"], T13/series_df["t3_mean"], label="sorting buckets")
    ax[i][j].plot(series_df["threads"], T14/series_df["t4_mean"], label="merging results")

    perfect_speedup = series_df["threads"].values
    ax[i][j].plot(perfect_speedup, perfect_speedup, label="perfect speedup", linestyle='--', color='black')

    ax[i][j].set_title(f"{series_nm} - Speedup")
    ax[i][j].legend()
    ax[i][j].set_xlabel("Number of Threads")
    ax[i][j].set_ylabel("Speedup")
    ax[i][j].grid()

    j += 1
    if j > 1:
        j = 0
        i += 1

plt.tight_layout()
```

```python
for series in bdf.groupby(["algo"]):
    series_df = series[1]
    series_nm = series[0][0]
 
    plt.plot(series_df["buckets"], series_df["t_mean"], label="total execution")
    # plt.plot(series_df["buckets"], series_df["t1_mean"], label="array fill")
    # plt.plot(series_df["buckets"], series_df["t2_mean"], label="distribution to buckets")
    # plt.plot(series_df["buckets"], series_df["t3_mean"], label="sorting buckets")
    # plt.plot(series_df["buckets"], series_df["t4_mean"], label="merging results")
    buckets_per_thread_suff = " Per Thread" if series_nm in ["sort_1"] else ""

    plt.title(f"{series_nm} - Execution Time vs Number of Buckets{buckets_per_thread_suff}")
    plt.xlabel(f"Number of Buckets{buckets_per_thread_suff}")
    
    plt.ylabel("Time [s]")
    plt.yscale("log")
    plt.xscale("log")
    plt.grid()

    min_index = series_df["t_mean"].idxmin()
    min_value = series_df["t_mean"].min()

    min_index_label = series_df["t_mean"].idxmin()
    min_value = series_df["t_mean"].min()
    min_index_pos = series_df.index.get_loc(min_index_label)
    try:
        min_buckets = series_df["buckets"].iloc[min_index_pos]
    except IndexError:
        min_buckets = series_df["buckets"].iloc[min_index_pos]

    plt.annotate(f"min: {min_value:.2f}s buckets: {min_buckets:.2e}", xy=(min_buckets, min_value), xytext=(min_buckets*0.05, min_value*1.5),
                     arrowprops=dict(facecolor='black', arrowstyle='->'))
    # plt.legend()

    plt.show()
```

```python
# Calculate the mean execution time for both algorithms
mean_bdf = bdf.groupby("buckets").agg(t_mean=("t_mean", "mean")).reset_index()

# Plot the mean execution time
plt.plot(mean_bdf["buckets"], mean_bdf["t_mean"], label="Mean Execution Time")

# Find the minimum value and its corresponding bucket
min_index = mean_bdf["t_mean"].idxmin()
min_value = mean_bdf["t_mean"].min()
min_buckets = mean_bdf["buckets"].iloc[min_index]

# Annotate the minimum value
plt.annotate(f"min: {min_value:.2f}s buckets: {min_buckets}", 
             xy=(min_buckets, min_value), 
             xytext=(min_buckets * 0.05, min_value * 1.5),
             arrowprops=dict(facecolor='black', arrowstyle='->'))

plt.title("Mean Execution Time vs Number of Buckets")
plt.xlabel("Number of Buckets")
plt.ylabel("Time [s]")
plt.yscale("log")
plt.xscale("log")
plt.grid()
plt.legend()
plt.show()
```

```python
for series in odf.groupby(["algo"]):
    series_df = series[1]
    series_nm = series[0]
    # Praca dla każdej kombinacji
    work = series_df["threads"] * series_df["t_mean"]
    series_df["work"] = work
    # Normalizacja względem pracy dla 1 wątku (i np. minimalnej liczby bucketów)
    ref = series_df[(series_df["threads"] == 1)]["work"].values
    if len(ref) > 0:
        series_df["work_norm"] = series_df["work"] / ref[0]
    else:
        series_df["work_norm"] = np.nan
    # Oś X: kombinacja bucketów/wątków
    x_labels = [f"{b}/{t}" for b, t in zip(series_df["buckets"], series_df["threads"])]
    plt.plot(x_labels, series_df["work_norm"], marker="o", label=f"{series_nm} work norm")
plt.axhline(1, color='gray', linestyle='--', label='ideal work')
plt.xlabel("Buckets/Threads")
plt.ylabel("Normalized Work")
plt.title("Normalized Work vs Buckets/Threads")
plt.xticks(rotation=45, ha="right")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
